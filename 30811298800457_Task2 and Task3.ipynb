```python
import pandas as pd 
df = pd.read_csv("Task1_combined_data.csv")
df.head()
```

```python
df.info()
```

```python
df["checkout_date"] = pd.to_datetime(df["checkout_date"])
df["return_date"] = pd.to_datetime(df["return_date"])
df["join_date"] = pd.to_datetime(df["join_date"])
df["grade"] = df["grade"].astype("Int64")
df["checkout_id"] = df["checkout_id"].astype("Int64")
df["publication_year"] = df["publication_year"].astype("Int64")

df.dtypes
```

```python
df.duplicated().sum()
```

```python
duplicates = df[df.duplicated(keep=False)]

duplicates
```

```python
df = df.drop_duplicates(keep='first')

df.duplicated().sum()
```

```python
df.isna().sum()
```

```python
missing_percentage = (df.isna().mean() * 100).round(2)

missing_percentage[missing_percentage > 0]
```

```python
# chechout_id 
max_id = df["checkout_id"].max()

missing_count = df["checkout_id"].isna().sum()

df.loc[df["checkout_id"].isna(), "checkout_id"] = range(
    max_id + 1,
    max_id + 1 + missing_count )

print (df["checkout_id"].isna().sum())
```

```python
# return_data 

# I will leave the missing values in the return_date column as they are because a missing return date
# does not necessarily indicate an error in the data. It can mean that the book has not been returned yet,
# so there is no return date to record. Therefore, filling these missing values with an estimated or
# arbitrary date could make the data inaccurate and misleading.
```

```python
# grade 

df["grade"] = df["grade"].fillna(df["grade"].mode()[0])

print (df["grade"].isna().sum())
```

```python
# join_date

df["join_date"] = df["join_date"].fillna(df["join_date"].mode()[0])

print (df["join_date"].isna().sum())
```

```python
# publication_year

df = df.dropna(subset=["publication_year"])

print (df["publication_year"].isna().sum())

# I decided to remove the rows with missing values in the publication_year column because I do not have
# a reliable source to determine the correct publication year for these books. Since the number of missing
# values is small, removing these rows is preferable to guessing or filling in incorrect information.
# This helps maintain the accuracy and reliability of the dataset.
```

```python
missing_percentage = (df.isna().mean() * 100).round(2)

missing_percentage[missing_percentage > 0]
```

```python
df = pd.read_csv("task2_cleaned_data.csv")

members_count = df.groupby("neighborhood")["member_id"].nunique()

checkouts_count = df.groupby("neighborhood").size()

fairness = pd.DataFrame({
    "members": members_count,
    "checkouts": checkouts_count })

print(fairness)
```

```python
fairness["member_share"] = (
    fairness["members"] / fairness["members"].sum())

fairness["checkout_share"] = (
    fairness["checkouts"] / fairness["checkouts"].sum())

fairness["difference"] = (
    fairness["checkout_share"] - fairness["member_share"])

print(fairness)
```
